Fazer um programa em Python que receba como entrada um arquivo contendo um texto qualquer e devolva como saída um arquivo de texto contendo o vocabulário (termos de indexação) de tal arquivo. O exemplo abaixo é ilustrativo; considere um texto com pelo menos 50 palavras. Indique na primeira célula de seu notebook qual a fonte de seu texto, se redigido pelo próprio grupo ou se a partir de outra fonte (sugestão de boa fonte de material: https://dominiopublico.mec.gov.br/).



````python
import re
import unicodedata # Importar o módulo unicodedata

def extract_vocabulary(input_filepath, output_filepath):
    """
    Extrai o vocabulário (termos de indexação) de um arquivo de texto
    e salva em outro arquivo.

    Args:
        input_filepath (str): Caminho para o arquivo de texto de entrada.
        output_filepath (str): Caminho para o arquivo de texto de saída do vocabulário.
    """
    try:
        with open(input_filepath, 'r', encoding='utf-8') as f:
            text = f.read()

        # 1. Converter para minúsculas
        text = text.lower()

        # 2. Remover acentuação usando unicodedata.normalize
        # NFKD transforma caracteres acentuados em sua forma base + acento separado
        # .encode('ascii', 'ignore') remove os acentos separados
        # .decode('utf-8') converte de volta para string
        text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8')

        # 3. Remover pontuação, números e outros caracteres não-alfabéticos
        # Substituir tudo que não é letra ou espaço por um espaço
        text = re.sub(r'[^a-z\s]', ' ', text)

        # 4. Dividir o texto em palavras
        words = text.split() # A função split() divide a string por espaços em branco
        words = [word.strip() for word in words if word.strip()] # Remover palavras vazias resultantes de múltiplos espaços

        # Verificar o número de palavras processadas
        if len(words) < 50:
            print(f"Atenção: O texto processado tem apenas {len(words)} palavras. O requisito é de pelo menos 50 palavras no texto de entrada.")

        # 5. Obter o vocabulário (palavras únicas) usando um set e ordenar
        vocabulary = sorted(list(set(words)))

        # 6. Escrever o vocabulário no arquivo de saída
        with open(output_filepath, 'w', encoding='utf-8') as f:
            for word in vocabulary:
                f.write(word + '\n')

        print(f"Vocabulário extraído com sucesso para '{output_filepath}'.")
        print(f"Total de palavras únicas: {len(vocabulary)}")

    except FileNotFoundError:
        print(f"Erro: O arquivo de entrada '{input_filepath}' não foi encontrado. Verifique o caminho e se o arquivo foi anexado corretamente.")
    except Exception as e:
        print(f"Ocorreu um erro: {e}")


# --- Exemplo de Uso ---

# 1. Definir o caminho para o arquivo de entrada fornecido pelo usuário.
# O arquivo a ser usado é o em anexo: /usr/Mãos DadasNão serei o poeta de um m.txt
input_filename = '/usr/Mãos DadasNão serei o poeta de um m.txt'

# 2. Definir o nome do arquivo de saída
output_filename = 'vocabulary.txt'

# 3. Chamar a função para extrair o vocabulário
extract_vocabulary(input_filename, output_filename)
````

```markdown
# Parte 2: Geração do Bag of Words (para ex2.ipynb)

Esta seção do código implementa a funcionalidade de "Bag of Words" (vetor de presença/ausência de termos do vocabulário) para múltiplos documentos de texto. As etapas incluem:

1.  Carregar o vocabulário gerado na Parte 1.
2.  Criar ou carregar documentos de texto de entrada.
3.  Processar cada documento para identificar a presença de termos do vocabulário.
4.  Apresentar o resultado como um vetor binário para cada documento.

**Instruções:**
*   Este código deve ser copiado para um novo notebook chamado `ex2.ipynb`.
*   Certifique-se de que o arquivo `vocabulary.txt` (gerado pelo `ex1.ipynb`) esteja no mesmo diretório ou que o caminho para ele seja especificado corretamente.
*   Os documentos de exemplo (`doc1.txt`, `doc2.txt`) serão criados automaticamente, mas você pode substituí-los pelos seus próprios arquivos.
```

````python
import re
import unicodedata
import os # Para listar arquivos em um diretório, se necessário

def clean_text(text):
    """
    Limpa o texto removendo pontuação, convertendo para minúsculas e removendo acentuação.
    """
    text = text.lower()
    text = unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8')
    text = re.sub(r'[^a-z\s]', ' ', text)
    return text

def generate_bag_of_words(document_filepath, vocabulary_list):
    """
    Gera o vetor 'bag of words' (presença/ausência) para um dado documento.

    Args:
        document_filepath (str): Caminho para o arquivo do documento.
        vocabulary_list (list): Lista de termos do vocabulário em ordem alfabética.

    Returns:
        list: Um vetor binário (0s e 1s) representando a presença/ausência de cada termo do vocabulário no documento.
    """
    bag_of_words = [0] * len(vocabulary_list)
    try:
        with open(document_filepath, 'r', encoding='utf-8') as f:
            text = f.read()

        processed_text = clean_text(text)
        words_in_document = set(processed_text.split())

        for i, term in enumerate(vocabulary_list):
            if term in words_in_document:
                bag_of_words[i] = 1
    except FileNotFoundError:
        print(f"Erro: Documento '{document_filepath}' não encontrado.")
    except Exception as e:
        print(f"Ocorreu um erro ao processar '{document_filepath}': {e}")

    return bag_of_words

# --- Exemplo de Uso para Bag of Words ---

# 1. Carregar o vocabulário a partir do arquivo 'vocabulary.txt'
vocabulary_file = 'vocabulary.txt'
vocabulary = []
try:
    with open(vocabulary_file, 'r', encoding='utf-8') as f:
        vocabulary = [line.strip() for line in f if line.strip()]
    print(f"Vocabulário carregado de '{vocabulary_file}'. Total de termos: {len(vocabulary)}")
except FileNotFoundError:
    print(f"Erro: O arquivo de vocabulário '{vocabulary_file}' não foi encontrado. Certifique-se de que o ex1.ipynb foi executado e gerou este arquivo.")
    exit()
except Exception as e:
    print(f"Ocorreu um erro ao carregar o vocabulário: {e}")
    exit()

# 2. Criar ou definir caminhos para os documentos de texto de entrada
# Para este exemplo, criaremos dois documentos menores.

doc1_content = """
Minha terra tem palmeiras, onde canta o sabiá.
As aves, que aqui gorjeiam, não gorjeiam como lá.
"""

doc2_content = """
Nosso céu tem mais estrelas, nossas várzeas têm mais flores.
Nossos bosques têm mais vida, nossa vida mais amores.
"""

document_paths = ['doc1.txt', 'doc2.txt']

# Escrever os documentos de exemplo em arquivos
with open('doc1.txt', 'w', encoding='utf-8') as f:
    f.write(doc1_content)
print("Documento de exemplo 'doc1.txt' criado.")

with open('doc2.txt', 'w', encoding='utf-8') as f:
    f.write(doc2_content)
print("Documento de exemplo 'doc2.txt' criado.")

# 3. Gerar e exibir o Bag of Words para cada documento
print("\n--- Bag of Words dos Documentos ---")
for doc_path in document_paths:
    bow = generate_bag_of_words(doc_path, vocabulary)
    print(f"Documento: {doc_path}")
    print(f"Bag of Words: {bow}\n")

# --- Dica para ler múltiplos arquivos de uma pasta (opcional) ---
# Se você tiver vários documentos em uma pasta, pode usar o seguinte:
# folder_path = 'caminho/para/seus/documentos/'
# all_document_paths = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.txt')]
# for doc_path in all_document_paths:
#    bow = generate_bag_of_words(doc_path, vocabulary)
#    print(f"Documento: {doc_path}")
#    print(f"Bag of Words: {bow}\n")
````
