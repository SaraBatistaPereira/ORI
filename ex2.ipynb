```python
import os
import re
import unicodedata

def limpar_texto(texto):

    # Remove acentos
    texto_sem_acento = unicodedata.normalize('NFKD', texto).encode('ASCII', 'ignore').decode('utf-8')
    # Converte para minúsculas
    texto_minusc = texto_sem_acento.lower()
    # Remove pontuação mantendo apenas letras e espaços em branco
    texto_limpo = re.sub(r'[^a-z\s]', '', texto_minusc)

    return texto_limpo

def extrair_palavras_unicas(texto):
    """Retorna um conjunto (set) com as palavras únicas de um texto limpo."""
    texto_limpo = limpar_texto(texto)
    # A função split() serva para separar o texto em um vetor de palavras
    return set(texto_limpo.split())

def gerar_e_salvar_vocabulario(diretorio_docs, caminho_saida_vocabulario):
    """
    Lê todos os documentos da pasta, gera o vocabulário unificado
    em ordem alfabética e salva em um arquivo .txt.
    """
    vocabulario_global = set()

    # Percorre todos os arquivos dentro do diretório
    for arquivo in os.listdir(diretorio_docs):
        if arquivo.endswith(".txt"):
            caminho_completo = os.path.join(diretorio_docs, arquivo)
            with open(caminho_completo, 'r', encoding='utf-8') as f:
                conteudo = f.read()
                # Atualiza o vocabulário global com as palavras deste documento
                vocabulario_global.update(extrair_palavras_unicas(conteudo))

    # Ordena o vetor de palavras alfabeticamente
    vocabulario_ordenado = sorted(list(vocabulario_global))

    # Salva o vocabulário gerado em disco
    with open(caminho_saida_vocabulario, 'w', encoding='utf-8') as f:
        for palavra in vocabulario_ordenado:
            f.write(palavra + '\n')

    return vocabulario_ordenado

def exibir_bag_of_words(diretorio_docs, caminho_vocabulario):
    """
    Lê o arquivo de vocabulário e os arquivos de texto para
    gerar a representação Bag of Words de cada documento.
    """
    # 1. Lê o vocabulário salvo
    with open(caminho_vocabulario, 'r', encoding='utf-8') as f:
        vocabulario = [linha.strip() for linha in f.readlines()]

    print(f"Vocabulário base ({len(vocabulario)} termos):")
    print(vocabulario, "\n")

    # 2. Processa cada documento da pasta para gerar o BoW
    for arquivo in sorted(os.listdir(diretorio_docs)):
        if arquivo.endswith(".txt"):
            caminho_completo = os.path.join(diretorio_docs, arquivo)
            with open(caminho_completo, 'r', encoding='utf-8') as f:
                conteudo = f.read()
                palavras_do_doc = extrair_palavras_unicas(conteudo)

                # Gera o vetor: 1 se a palavra do vocabulário está no documento, senão 0
                bag_of_words = [1 if termo in palavras_do_doc else 0 for termo in vocabulario]

                print(f"Documento: {arquivo}")
                print(f"Bag of Words:\n{bag_of_words}\n")

DIRETORIO = "docs_ex2"
ARQUIVO_VOCABULARIO = "vocabulario_ex2.txt"

print("--- ETAPA 1: Gerando Vocabulário a partir dos documentos ---")
gerar_e_salvar_vocabulario(DIRETORIO, ARQUIVO_VOCABULARIO)
print(f"Vocabulário criado e salvo em: '{ARQUIVO_VOCABULARIO}'\n")

print("--- ETAPA 2: Lendo arquivos e gerando Bag of Words ---")
exibir_bag_of_words(DIRETORIO, ARQUIVO_VOCABULARIO)
```
