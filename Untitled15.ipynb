````python
import os
import re
import time
import math
import unicodedata
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# Criando o diretório para armazenar os arquivos de texto das músicas
musicas_dir = './clube_da_esquina_letras'
os.makedirs(musicas_dir, exist_ok=True)

# Dicionário com as letras das principais faixas do álbum (excluindo instrumentais e 'Dos Cruces')
# Apenas versos estão inclusos, sem os títulos no corpo do texto
letras_album = {
    "Tudo_Que_Voce_Podia_Ser.txt": """Com sol e chuva você sonhava
Com o que queria ser
Depois de se perder
Mas hoje eu já não sei quem é você
Só sei que você não é o que pensou
Que tudo aquilo se perdeu
O que você podia ser""",

    "Cais.txt": """Para quem quer se soltar
Invento o cais
Invento mais que a solidão se deixa
Invento a lua nova a clarear
Invento o amor e sei a dor de encontrar
Eu queria ser feliz
Invento o mar
Invento o cais""",

    "O_Trem_Azul.txt": """Peguei um trem azul
Com o sol de amanhã
No vento os meus cabelos
E a luz na minha janela
Não quero mais o que já passou
Não quero mais o que já perdi
Quero o que está por vir""",

    "Clube_da_Esquina_No_2.txt": """Porque se chamavam homens
Também se chamavam sonhos
E os sonhos cresciam
Em pleno sol da manhã
E no vento que vinha das minas
E no asfalto quente da cidade
Eles eram muitos e hoje são só lembranças""",

    "Cravo_e_Canela.txt": """Cravo e canela na pele
O cheiro de mato e de flor
Sabor de mel na boca
O sol brilhando no seu olhar
Seu corpo moreno correndo na areia
Vem me dar seu amor""",

    "Paisagem_da_Janela.txt": """Da janela lateral do quarto de dormir
Vejo uma igreja um sinal de glória
Vejo um muro branco e um voo de pássaro
Vejo uma grade um velho sinal
Mensageiro natural de coisas de outro dia
Quero contar de coisas de agora""",

    "Um_Girassol_da_Cor_Seu_Cabelo.txt": """Vento solar e estrelas do mar
Um girassol da cor de seu cabelo
Se eu não puder te ver hoje
Não quero ver mais nada
O mundo gira e tudo muda
Mas o meu amor por você continua""",

    "San_Vicente.txt": """Coração americano
Um farol na noite escura
Um porto de chegada
Uma canção na América do Sul
As estradas poeirentas
Onde os homens caminham livremente""",

    "O_Trem_da_Central.txt": """O trem da central que parte toda hora
Leva gente que trabalha e chora
Leva sonhos e cansaço
Na velocidade do aço
De volta para o subúrbio e para o lar""",

    "Nada_Sera_Como_Antes.txt": """Eu já sei que nada será como antes amanhã
Que as coisas mudam e que o tempo passa
Mas eu guardo essa certeza
No peito e na alma
Que tudo que fomos ainda vive em nós""",

    "Estrelas.txt": """Estrelas no céu a brilhar
Mostrando o caminho para quem quer amar
A noite é nossa guia
E a música nos leva além""",

    "Ao_Que_Vai_Nascer.txt": """Para o novo dia que vai nascer
Trago no peito a esperança de ver
Um mundo sem guerras e sem dor
Onde a lei maior seja o amor"""
}

# Gravando as letras em arquivos .txt
for nome_arquivo, conteudo in letras_album.items():
    caminho = os.path.join(musicas_dir, nome_arquivo)
    with open(caminho, 'w', encoding='utf-8') as f:
        f.write(conteudo)

print(f"Arquivos criados com sucesso no diretório: {musicas_dir}")
````

```markdown
## Funções para Tratamento de Texto e Cálculo do TF-IDF (com Log na Base 2)
```

````python
def remover_acentos_e_pontuacao(texto):
    # Normaliza o texto para remover acentos
    texto_normalizado = unicodedata.normalize('NFKD', texto)
    texto_sem_acentos = "".join([c for c in texto_normalizado if not unicodedata.combining(c)])
    # Remove pontuação mantendo apenas letras e espaços, e converte para minúsculas
    texto_limpo = re.sub(r'[^a-zA-Z\s]', '', texto_sem_acentos)
    return texto_limpo.lower()

def processar_documento(caminho_arquivo):
    with open(caminho_arquivo, 'r', encoding='utf-8') as f:
        conteudo = f.read()
    texto_limpo = remover_acentos_e_pontuacao(conteudo)
    # Retorna a lista de termos (palavras)
    return [palavra for palavra in texto_limpo.split() if palavra]
````

```markdown
## Execução do Exercício 2: Respostas aos Itens
```

````python
# c) Medindo o tempo para criar o vocabulário
tempo_inicio_vocab = time.time()

arquivos = [f for f in os.listdir(musicas_dir) if f.endswith('.txt')]
docs_termos = {}
vocabulario_set = set()

for arq in arquivos:
    caminho = os.path.join(musicas_dir, arq)
    termos = processar_documento(caminho)
    docs_termos[arq] = termos
    vocabulario_set.update(termos)

vocabulario_ordenado = sorted(list(vocabulario_set))

# Salvar vocabulário em arquivo de texto
with open('vocabulario.txt', 'w', encoding='utf-8') as f:
    for termo in vocabulario_ordenado:
        f.write(f"{termo}\n")

tempo_fim_vocab = time.time()
tempo_vocab = tempo_fim_vocab - tempo_inicio_vocab

# d) Medindo o tempo para calcular o TF-IDF da coleção inteira
tempo_inicio_tfidf = time.time()

N = len(arquivos)

# Cálculo do IDF (log base 2)
df_termos = {termo: 0 for termo in vocabulario_ordenado}
for termo in vocabulario_ordenado:
    for arq, termos in docs_termos.items():
        if termo in termos:
            df_termos[termo] += 1

idf = {}
for termo, freq_doc in df_termos.items():
    # idf = log2(N / df)
    idf[termo] = math.log2(N / freq_doc)

# Cálculo do TF e TF-IDF para cada par Termo-Documento
tfidf_matriz = {}
for arq, termos in docs_termos.items():
    tfidf_matriz[arq] = {}
    total_termos = len(termos)
    # Frequência de cada termo no doc atual
    freq_termos_doc = {}
    for t in termos:
        freq_termos_doc[t] = freq_termos_doc.get(t, 0) + 1

    for termo in vocabulario_ordenado:
        tf = freq_termos_doc.get(termo, 0) / total_termos if total_termos > 0 else 0
        tfidf_matriz[arq][termo] = tf * idf[termo]

tempo_fim_tfidf = time.time()
tempo_tfidf = tempo_fim_tfidf - tempo_inicio_tfidf

# Encontrando o maior peso TF-IDF
maior_peso = -1
termo_maior = ""
doc_maior = ""

for arq, termos_pesos in tfidf_matriz.items():
    for termo, peso in termos_pesos.items():
        if peso > maior_peso:
            maior_peso = peso
            termo_maior = termo
            doc_maior = arq

print("=== RESPOSTAS EXERCÍCIO 2 ===\n")
print(f"a) Tamanho do vocabulário da coleção: {len(vocabulario_ordenado)} termos.")
print(f"b) Termo com maior peso TF-IDF: '{termo_maior}' no documento '{doc_maior}' com peso {maior_peso:.6f}")
print(f"c) Tempo gasto para criar o vocabulário: {tempo_vocab:.6f} segundos.")
print(f"d) Tempo gasto para calcular o TF-IDF da coleção: {tempo_tfidf:.6f} segundos.")
````

```markdown
## Exercício 3 - Planilha de Termos por Documentos e Gráficos
```

````python
# a) Criando a matriz de termos por documentos (DataFrame) e exportando para .xlsx
matriz_df = pd.DataFrame(index=vocabulario_ordenado)
for arq in sorted(arquivos):
    coluna_valores = []
    for termo in vocabulario_ordenado:
        coluna_valores.append(tfidf_matriz[arq][termo])
    matriz_df[arq] = coluna_valores

# Salvar no formato Excel xlsx
nome_planilha = "matriz_termos_documentos.xlsx"
matriz_df.to_excel(nome_planilha)
print(f"Planilha '{nome_planilha}' gerada com sucesso!")
````

````python
# b) Criar os gráficos de TF vs. IDF e o de TF-IDF

# 1. Calcular TF para cada termo baseado na coleção inteira
total_palavras_colecao = sum(len(t) for t in docs_termos.values())
tf_colecao = {}

# Contagem total de cada termo na coleção
contagem_total = {termo: 0 for termo in vocabulario_ordenado}
for termos in docs_termos.values():
    for t in termos:
        contagem_total[t] += 1

for termo in vocabulario_ordenado:
    tf_colecao[termo] = contagem_total[termo] / total_palavras_colecao

# 2. Criar estrutura de dados para plotagem
data_plot = []
for termo in vocabulario_ordenado:
    # TF-IDF médio ou máximo para representação no gráfico (usaremos o máximo obtido na coleção)
    max_tfidf = max(tfidf_matriz[arq][termo] for arq in arquivos)
    data_plot.append({
        'termo': termo,
        'TF': tf_colecao[termo],
        'IDF': idf[termo],
        'TF-IDF': max_tfidf
    })

df_plot = pd.DataFrame(data_plot)

# Ordenar termos de forma decrescente de TF
df_plot = df_plot.sort_values(by='TF', ascending=False).reset_index(drop=True)

# Gerar os gráficos e exportar para PDF
pdf_filename = 'graficos_ponderacao.pdf'

with PdfPages(pdf_filename) as pdf:
    # Gráfico 1: TF vs IDF
    fig1, ax1 = plt.subplots(figsize=(12, 6))
    ax1.plot(df_plot.index, df_plot['TF'], label='TF (Coleção)', color='blue', marker='o', markersize=3, linestyle='-')
    ax1.plot(df_plot.index, df_plot['IDF'], label='IDF', color='green', marker='s', markersize=3, linestyle='--')
    ax1.set_yscale('log') # Eixo Y em escala logarítmica (semi-log)
    ax1.set_title('Gráfico TF vs IDF (Ordenado por TF Decrescente) - Escala Semi-Log')
    ax1.set_xlabel('Índice do Termo (Ordenado)')
    ax1.set_ylabel('Valor (Escala Log)')
    ax1.grid(True, which="both", ls="--", alpha=0.5)
    ax1.legend()
    plt.tight_layout()
    pdf.savefig(fig1)
    plt.show()
    plt.close(fig1)

    # Gráfico 2: TF-IDF
    fig2, ax2 = plt.subplots(figsize=(12, 6))
    ax2.plot(df_plot.index, df_plot['TF-IDF'], label='TF-IDF Max', color='red', marker='^', markersize=3, linestyle='-')
    ax2.set_yscale('log') # Eixo Y em escala logarítmica (semi-log)
    ax2.set_title('Gráfico de TF-IDF Máximo (Mantendo Ordem do TF) - Escala Semi-Log')
    ax2.set_xlabel('Índice do Termo (Ordenado por TF Decrescente)')
    ax2.set_ylabel('TF-IDF (Escala Log)')
    ax2.grid(True, which="both", ls="--", alpha=0.5)
    ax2.legend()
    plt.tight_layout()
    pdf.savefig(fig2)
    plt.show()
    plt.close(fig2)

print(f"Gráficos exportados com sucesso para o arquivo PDF: '{pdf_filename}'")
````
