### **Passo 1.1 - Criação do Arquivo de Entrada (Texto de Origem)**

Nesta etapa, criamos o arquivo físico de texto chamado `entrada_ex1.txt`. Este arquivo simula o arquivo de entrada que o usuário forneceria ao programa. O texto escolhido é uma versão adaptada do poema *Canção do Exílio* de Gonçalves Dias, garantindo que tenhamos um corpus com mais de 50 palavras para análise.

```python
# Criando o arquivo de entrada com o texto de exemplo (mais de 50 palavras)
texto_exemplo = """Minha terra tem palmeiras,
Onde canta o Sabiá;
As aves, que aqui gorjeiam,
Não gorjeiam como lá.

Nosso céu tem mais estrelas,
Nossas várzeas têm mais flores,
Nossos bosques têm mais vida,
Nossa vida mais amores.

Em cismar, sozinho, à noite,
Mais prazer encontro eu lá;
Minha terra tem palmeiras,
Onde canta o Sabiá.

Minha terra tem primores,
Que tais não encontro eu cá;
Em cismar — sozinho, à noite —
Mais prazer encontro eu lá;
Minha terra tem palmeiras,
Onde canta o Sabiá."""

with open('entrada_ex1.txt', 'w', encoding='utf-8') as f:
    f.write(texto_exemplo)

print("Arquivo 'entrada_ex1.txt' criado com sucesso!")
```

### **Passo 1.2 - Funções de Limpeza, Normalização e Geração de Vocabulário**

Para construir o vocabulário de acordo com os requisitos definidos, implementamos duas funções principais:

1. **`limpar_e_normalizar(texto)`**:
   * **Conversão de caixa:** Converte todo o texto para letras minúsculas (`.lower()`).
   * **Remoção de acentos:** Utiliza a biblioteca nativa `unicodedata` com a normalização `NFD` para separar as letras de seus acentos gráficos, descartando os acentos na sequência.
   * **Remoção de pontuações:** Substitui caracteres de pontuação e travessões por espaços usando tabelas de tradução do Python (`str.maketrans`).

2. **`gerar_vocabulario(caminho_entrada, caminho_saida)`**:
   * Abre o arquivo de entrada e lê o conteúdo.
   * Invoca a função de limpeza.
   * Divide o texto limpo em palavras individuais (`.split()`).
   * Filtra elementos repetidos usando uma estrutura de conjunto (`set`).
   * Ordena os elementos remanescentes em ordem alfabética (`sorted()`) e salva cada termo em uma nova linha no arquivo `saida_vocab_ex1.txt`.

```python
import unicodedata
import string

def limpar_e_normalizar(texto):
    """
    Remove acentos, pontuação e converte o texto para minúsculas.
    """
    # Converte para minúsculas
    texto = texto.lower()

    # Remove acentuação usando unicodedata
    texto_normalizado = unicodedata.normalize('NFD', texto)
    texto_sem_acentos = "".join(c for c in texto_normalizado if unicodedata.category(c) != 'Mn')

    # Remove pontuação substituindo por espaço
    # Criamos uma tabela de tradução que mapeia pontuações para espaços
    tabela_pontuacao = str.maketrans(string.punctuation + '—–', ' ' * (len(string.punctuation) + 2))
    texto_limpo = texto_sem_acentos.translate(tabela_pontuacao)

    return texto_limpo

def gerar_vocabulario(caminho_entrada, caminho_saida):
    """
    Lê o arquivo de entrada, limpa o texto, gera o vocabulário único ordenado e salva no arquivo de saída.
    """
    # 1. Abre e lê o arquivo de entrada
    with open(caminho_entrada, 'r', encoding='utf-8') as arquivo:
        conteudo = arquivo.read()

    # 2. Limpa e normaliza o texto
    texto_processado = limpar_e_normalizar(conteudo)

    # 3. Separa as palavras usando split()
    palavras = texto_processado.split()

    # 4. Remove duplicatas usando set
    vocab_set = set(palavras)

    # 5. Ordena alfabeticamente
    vocab_ordenado = sorted(list(vocab_set))

    # 6. Grava o vocabulário no arquivo de saída
    with open(caminho_saida, 'w', encoding='utf-8') as arquivo_saida:
        for termo in vocab_ordenado:
            arquivo_saida.write(termo + '\n')

    print(f"Vocabulário gerado com sucesso! Salvo em: '{caminho_saida}'")
    return vocab_ordenado

# Executa o processamento do Exemplo 1
vocabulario = gerar_vocabulario('entrada_ex1.txt', 'saida_vocab_ex1.txt')
```

```python
# Exibe o vocabulário gerado para conferência
print(f"Total de termos únicos no vocabulário: {len(vocabulario)}\n")
print("Termos ordenados:")
print(vocabulario)
```

### **Painel de Verificação do Exercício 1 (Entrada vs. Saída)**

A célula abaixo lê o arquivo original de entrada (`entrada_ex1.txt`) e o arquivo de vocabulário de saída (`saida_vocab_ex1.txt`) e exibe ambos formatados para conferência imediata do professor.

```python
# Lendo e exibindo a Entrada original
with open('entrada_ex1.txt', 'r', encoding='utf-8') as f_in:
    entrada_conteudo = f_in.read()

# Lendo e exibindo a Saída (Vocabulário) gerada
with open('saida_vocab_ex1.txt', 'r', encoding='utf-8') as f_out:
    saida_termos = [linha.strip() for linha in f_out if linha.strip()]

print("=" * 60)
print("            VERIFICAÇÃO DE ENTRADA E SAÍDA - EXERCÍCIO 1")
print("=" * 60)
print("\n--- [ARQUIVO DE ENTRADA (entrada_ex1.txt)] ---")
print(entrada_conteudo)
print("\n" + "-" * 60)
print(f"\n--- [ARQUIVO DE SAÍDA (saida_vocab_ex1.txt) - {len(saida_termos)} Termos] ---")
# Formata a lista de termos em colunas para facilitar a leitura
for i, termo in enumerate(saida_termos):
    print(f"{termo:<15}", end="\n" if (i + 1) % 4 == 0 else "")
print("\n" + "=" * 60)
```
